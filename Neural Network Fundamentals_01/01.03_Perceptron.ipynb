# Introduction

The Perceptron was proposed by Frank Rosenblatt in 1958.

It is the first learning algorithm for artificial neurons.

Unlike the McCulloch-Pitts neuron, the Perceptron can learn its weights from data.

---

# Linear Neuron

A Perceptron computes a weighted sum of inputs.

Mathematical model:

```text
z = w1x1 + w2x2 + ... + wnxn + b
```

Where:

* x = inputs
* w = weights
* b = bias

---

# Perceptron Architecture

```text
x1 ----\
         \
x2 ------>( Σ ) ---> Activation ---> Output
         /
x3 ----/
```

---

# Python Example: Linear Neuron

```python
def linear_neuron(inputs, weights, bias):

    z = sum(i*w for i,w in zip(inputs,weights)) + bias

    return z


inputs = [2,3]
weights = [0.5,0.8]
bias = 1

print(linear_neuron(inputs,weights,bias))
```

---

# Activation Function

The activation function determines whether the neuron fires.

Perceptrons use the Step Function.

```text
Output = 1 if z ≥ 0
Output = 0 if z < 0
```

---

# Python Implementation

```python
def step_function(z):

    if z >= 0:
        return 1
    else:
        return 0
```

---

# Complete Perceptron

```python
def perceptron(inputs,weights,bias):

    z = sum(i*w for i,w in zip(inputs,weights)) + bias

    return step_function(z)


print(perceptron([1,1],[1,1],-1.5))
```

---

# Decision Boundary

A decision boundary separates classes.

For two inputs:

```text
w1x1 + w2x2 + b = 0
```

This equation represents a line.

Points on one side:

Class 0

Points on the other side:

Class 1

---

# Visualizing Decision Boundary

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5,5,100)

y = -(2*x+1)/3

plt.plot(x,y)
plt.grid()
plt.show()
```
