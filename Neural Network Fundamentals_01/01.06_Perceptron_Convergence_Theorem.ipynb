## Learning Objectives

* Understand convergence.

---

# What is Convergence?

Convergence means:

```text
Training eventually stops making mistakes.
```

Weights stabilize.

---

# Perceptron Convergence Theorem

Statement:

If training data is linearly separable, the Perceptron Learning Algorithm is guaranteed to find a solution in a finite number of updates.

---

# Conditions

The theorem holds when:

1. Dataset is linearly separable.
2. Learning rate is positive.
3. Training samples are finite.

---

# Why Does It Converge?

Each update moves weights closer to the correct separating hyperplane.

Eventually:

```text
All points become correctly classified.
```

Error becomes zero.

Training stops.

---

# Geometric Interpretation

```text
Wrong Boundary
      ↓

Weight Updates
      ↓

Better Boundary
      ↓

Perfect Boundary
```

---

# Demonstration

```python
from sklearn.linear_model import Perceptron
from sklearn.datasets import make_classification

X,y = make_classification(
    n_samples=100,
    n_features=2,
    n_redundant=0,
    n_clusters_per_class=1,
    random_state=42
)

model = Perceptron()

model.fit(X,y)

print(model.score(X,y))
```

Expected output:

```text
1.0
```

---

# When Convergence Fails

Perceptron cannot converge when:

* Data is noisy
* Data is not linearly separable
* XOR-like patterns exist

Example:

```text
XOR
```

No separating hyperplane exists.

---

# Limitations

Perceptrons:

✅ AND

✅ OR

❌ XOR

❌ Complex patterns

This led to:

* Multi-Layer Perceptrons
* Backpropagation
* Deep Learning

---

# Key Takeaways

* Perceptron converges only for linearly separable datasets.
* Convergence is guaranteed mathematically.
* XOR breaks the theorem.
* Multilayer networks solve this limitation.
