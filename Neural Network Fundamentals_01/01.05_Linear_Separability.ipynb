# What is Linear Separability?

A dataset is linearly separable if a straight line can separate classes.

Example:

```text
Class A ●
Class B ○
```

Can be separated by one line.

---

# Hyperplane

General decision equation:

```text
wᵀx + b = 0
```

Dimensions:

| Features | Hyperplane |
| -------- | ---------- |
| 2        | Line       |
| 3        | Plane      |
| n        | Hyperplane |

---

# Example Dataset

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([
    [1,1],
    [2,2],
    [4,4],
    [5,5]
])

y = [0,0,1,1]

for point,label in zip(X,y):

    if label==0:
        plt.scatter(point[0],point[1])
    else:
        plt.scatter(point[0],point[1],marker='x')

plt.show()
```

---

# Decision Boundary

A decision boundary is the surface separating classes.

For 2D:

```text
Line
```

For 3D:

```text
Plane
```

For higher dimensions:

```text
Hyperplane
```

# XOR Dataset

```text
(0,0)
(1,1)

Class 0


(0,1)
(1,0)

Class 1
```

Cannot be separated using a single line.

---

# Key Takeaways

* Perceptrons require linear separability.
* Hyperplanes separate classes.
* XOR is not linearly separable.
