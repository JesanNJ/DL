# 01.04 Perceptron Learning Rule

## Learning Objectives

* Understand weight updates.
* Understand learning rate.

---

# Why Learning is Needed

Initially weights are random.

The perceptron must adjust weights based on mistakes.

---

# Error

```text
Error = Target - Prediction
```

Example:

```text
Target = 1
Prediction = 0

Error = 1
```

---

# Weight Update Rule

Perceptron Learning Rule:

```text
w_new = w_old + η × error × x
```

Where:

* η = learning rate
* x = input

---

# Learning Rate

Learning rate controls update size.

Small η:

```text
Slow learning
```

Large η:

```text
Fast but unstable learning
```

Typical values:

```python
0.001
0.01
0.1
```

---

# Example Update

```python
w = 0.5

eta = 0.1

error = 1

x = 2

w_new = w + eta*error*x

print(w_new)
```

---

# Complete Training Example

```python
X = [
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

y = [0,0,0,1]

weights = [0.0,0.0]
bias = 0

eta = 0.1

for epoch in range(10):

    for inputs,target in zip(X,y):

        z = sum(i*w for i,w in zip(inputs,weights)) + bias

        pred = 1 if z >= 0 else 0

        error = target - pred

        for i in range(len(weights)):
            weights[i] += eta*error*inputs[i]

        bias += eta*error

print(weights,bias)
```

---

# Key Takeaways

* Perceptron learns from errors.
* Weights change after mistakes.
* Learning rate controls update size.
