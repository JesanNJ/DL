# Introduction

The biological neuron is the fundamental building block of the nervous system. It is a specialized cell responsible for receiving, processing, and transmitting information through electrical and chemical signals.

The concept of artificial neural networks (ANNs) is inspired by the structure and functioning of biological neurons. Although artificial neurons are much simpler than biological neurons, they are based on the same fundamental idea of receiving inputs, processing them, and generating outputs.

---

# Structure of a Biological Neuron

A typical biological neuron consists of four major parts:

1. Dendrites
2. Cell Body (Soma)
3. Axon
4. Synapse

```text
           Dendrites
          /  |  |  \
         /   |  |   \
        /    |  |    \
     --------------------
     |       Soma       |
     |   (Cell Body)    |
     --------------------
               |
               |
               |
             Axon
               |
               |
               |
        ----------------
        |   Synapse    |
        ----------------
               |
               |
       Next Neuron
```

# Human Neuron

A human neuron is a highly specialized cell that can communicate with thousands of other neurons.

The human brain contains approximately:

* 86 billion neurons
* Trillions of synaptic connections

Each neuron can receive signals from many other neurons, process the information, and transmit signals to other neurons.

### Main Functions

* Receive information
* Process information
* Transmit information
* Store learned patterns

---

# Dendrite

## Definition

Dendrites are branch-like extensions that receive signals from other neurons.

### Functions

* Collect incoming signals
* Carry signals toward the cell body
* Increase the surface area available for communication

### Characteristics

* Short and highly branched
* Receive electrical and chemical signals
* Connected to synapses

### Analogy

Think of dendrites as the input channels of a computer system.

```text
Input Signals
      ↓
  Dendrites
      ↓
     Soma
```

---

# Cell Body (Soma)

## Definition

The cell body, also called the soma, is the central processing unit of the neuron.

### Functions

* Processes incoming signals
* Contains the nucleus
* Maintains cell health
* Decides whether to generate an output signal

### Analogy

The soma acts similarly to a processor (CPU) in a computer.

```text
Inputs
  ↓
Soma
  ↓
Decision
```

---

# Axon

## Definition

The axon is a long fiber that carries electrical impulses away from the cell body.

### Functions

* Transmits signals to other neurons
* Conducts action potentials
* Enables long-distance communication

### Characteristics

* Usually longer than dendrites
* Covered by myelin sheath in many neurons
* Ends at axon terminals

### Analogy

The axon acts like a communication cable carrying data from one device to another.

```text
Soma
  ↓
Axon
  ↓
Output Signal
```

---

# Synapse

## Definition

A synapse is the junction between two neurons where communication occurs.

### Components

1. Presynaptic neuron
2. Synaptic cleft
3. Postsynaptic neuron

```text
Neuron A
   |
   |
[Axon Terminal]
       ↓
------------------
 Synaptic Cleft
------------------
       ↓
[Dendrite]
   |
Neuron B
```

### Functions

* Transfers information between neurons
* Uses neurotransmitters
* Enables learning and memory

### Common Neurotransmitters

* Dopamine
* Serotonin
* Acetylcholine
* Glutamate
* GABA

---

# Signal Transmission in a Neuron

The flow of information inside a neuron follows this path:

```text
Dendrites
    ↓
Cell Body (Soma)
    ↓
Axon
    ↓
Synapse
    ↓
Next Neuron
```

### Step-by-Step Process

1. Dendrites receive signals.
2. Signals travel to the soma.
3. Soma processes the signals.
4. If the signal is strong enough, an action potential is generated.
5. The action potential travels along the axon.
6. Neurotransmitters are released at the synapse.
7. The next neuron receives the signal.

# Biological Neuron vs Artificial Neuron

| Biological Neuron             | Artificial Neuron           |
| ----------------------------- | --------------------------- |
| Dendrites                     | Inputs                      |
| Synapses                      | Weights                     |
| Soma                          | Summation Function          |
| Activation Process            | Activation Function         |
| Axon                          | Output                      |
| Learning via Synaptic Changes | Learning via Weight Updates |
